La primera línea genera un número aleatorio entre 0 y 1 para cada uno de los cuatro genes. Compara cada número con 0.5 y produce una lista de valores lista de valores booleanos (True o False), dónde True seleccionará el gen del primer padre y False el gen del segundo. 

La segunda línea, np.where construye al hijo seleccionado, posición por posición, el gen del primer padre cuando la condición es verdadera y el del segundo cuando es falsa. Cada gen tiene aproximadamente un 50% de probabilidad de proceder de cualquiera de los dos padres independientemente de su posición. Por ejemplo, si la máscara aleatoria fuera  [True False True False], con los padres (0,0,0,0) y (8,8,6,8) el hijo sería (0,8,0,8).

Así np.where seleccionaría todos los genes del primer padre, el hijo sería una copia del padre y ya no existiría una cruza de información de ambos padres, por lo que se reduciría la diversidad genética.
